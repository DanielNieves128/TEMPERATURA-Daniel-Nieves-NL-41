Fórmulas de ConversiónPara convertir temperaturas entre estas dos escalas, utilizamos las siguientes relaciones matemáticas:De Grados Celsius (°C) a Grados Fahrenheit (°F):$$°F = (°C \times \frac{9}{5}) + 32$$(O también: $°F = (°C \times 1.8) + 32$)De Grados Fahrenheit (°F) a Grados Celsius (°C):$$°C = \frac{(°F - 32) \times 5}{9}$$(O también: $°C = \frac{°F - 32}{1.8}$)
2. Ejemplos Resueltos Paso a PasoEjemplo A: Convertir 25 °C a °F
* Identificar la fórmula: $°F = (°C \times 1.8) + 32$
* Sustituir el valor: $°F = (25 \times 1.8) + 32$
* Multiplicar: $25 \times 1.8 = 45$
* Sumar 32: $45 + 32 = 77$
* Resultado: 25 °C equivalen a 77 °F.

Ejemplo B: Convertir 95 °F a °C
* Identificar la fórmula: $°C = (°F - 32) / 1.8$
* Sustituir el valor: $°C = (95 - 32) / 1.8$
* Restar 32: $95 - 32 = 63$
* Dividir entre 1.8: $63 / 1.8 = 35$
* Resultado: 95 °F equivalen a 35 °C.
3. Código en PythonAquí tienes el programa interactivo siguiendo tus requerimientos específicos. He nombrado las funciones de conversión como convertir_CF (Celsius a Fahrenheit) y convertir_FC (Fahrenheit a Celsius) para mayor claridad.

```python
def convertir_CF():
    """Convierte de Celsius a Fahrenheit"""
    try:
        celsius = float(input("Ingresa los grados Celsius: "))
        fahrenheit = (celsius * 9/5) + 32
        print(f"\n>>> {celsius}°C es igual a {fahrenheit:.2f}°F\n")
    except ValueError:
        print("Error: Por favor ingresa un número válido.")

def convertir_FC():
    """Convierte de Fahrenheit a CelsiusSuffix"""
    try:
        fahrenheit = float(input("Ingresa los grados Fahrenheit: "))
        celsius = (fahrenheit - 32) * 5/9
        print(f"\n>>> {fahrenheit}°F es igual a {celsius:.2f}°C\n")
    except ValueError:
        print("Error: Por favor ingresa un número válido.")

def menu():
    while True:
        print("--- MENÚ DE CONVERSIONES ---")
        print("1. Convertir grados Centígrados a Fahrenheit")
        print("2. Convertir grados Fahrenheit a Centígrados")
        print("3. Salir")

        opcion = input("Selecciona una opción (1-3): ")

        if opcion == '1':
            convertir_CF()
        elif opcion == '2':
            convertir_FC()
        elif opcion == '3':
            print("Saliendo del programa ... ¡Hasta luego!")
            print("Daniel Nieves NL = 41")
            break
        else:
            print("Opción no válida, intenta de nuevo.\n")

# Ejecución del programa
if __name__ == "__main__":
    menu()
```
